```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

X_train = pd.read_csv("../data/processed/X_train.csv")
X_test = pd.read_csv("../data/processed/X_test.csv")
y_train = pd.read_csv("../data/processed/y_train.csv").squeeze()
y_test = pd.read_csv("../data/processed/y_test.csv").squeeze()
print(X_train.shape)  # should be 63 columns
```

```python
# Look BEFORE clipping
sns.boxplot(x=y_train, y=X_train["Attr1"])
plt.title("Attr1 before clipping")
plt.show()
```

```python
# Learn the limits from train only
low = X_train.quantile(0.01)
high = X_train.quantile(0.99)

X_train_clip = X_train.clip(lower=low, upper=high, axis=1)
X_test_clip = X_test.clip(lower=low, upper=high, axis=1)
```

```python
# Look AFTER clipping
sns.boxplot(x=y_train, y=X_train_clip["Attr1"])
plt.title("Attr1 after clipping")
plt.show()
```

```python
# Save for the next step
X_train_clip.to_csv("../data/processed/X_train_clip.csv", index=False)
X_test_clip.to_csv("../data/processed/X_test_clip.csv", index=False)
```
